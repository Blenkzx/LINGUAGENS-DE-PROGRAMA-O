```python
import numpy as np
```

## Nível 1: Básico — Faturamento Semanal e Filtros

### Passo 1: Crie um array unidimensional `pecas_vendidas`.

```python
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
display(pecas_vendidas)
```

### Passo 2: Calcule o faturamento semanal, sabendo que o preço médio de cada peça é de R$ 50,00.

```python
preco_peca = 50
faturamento_semanal = pecas_vendidas * preco_peca
display(faturamento_semanal)
```

### Passo 3: Identifique os dias de pico (vendas estritamente mais de 200 peças).

```python
dias_pico_mascara = pecas_vendidas > 200
dias_pico = pecas_vendidas[dias_pico_mascara]
display(dias_pico_mascara)
display(f"Vendas nos dias de pico: {dias_pico}")
```

## Nível 2: Intermediário — Múltiplas Filiais e Falhas de Sistema

### Passo 1: Crie uma matriz 2D para vendas de três filiais.

```python
vendas_filiais = np.array([
    [200, 220, np.nan, 250],  # Loja A
    [150, 180, 160, 190],   # Loja B
    [300, 310, 290, 330]    # Loja C
])
display(vendas_filiais)
```

### Passo 2: Calcule o total de peças vendidas por cada loja (agregando pelas linhas).

```python
total_pecas_por_loja = np.nansum(vendas_filiais, axis=1)
display(total_pecas_por_loja)
```

### Passo 3: Calcule a média diária geral da rede, ignorando valores `nan`.

```python
media_diaria_geral = np.nanmean(vendas_filiais)
display(f"Média diária geral da rede (ignorando NaN): {media_diaria_geral:.2f}")
```

### Passo 4: Crie uma nova matriz condicional usando `np.where()` para 'Meta Atingida' ou 'Abaixo'.

```python
meta_diaria = 200
status_vendas = np.where(vendas_filiais >= meta_diaria, 'Meta Atingida', 'Abaixo')
display(status_vendas)
```

## Nível 3: Avançado — Categorização e Destaques de Marketing

### Passo 1: Crie uma amostra com 50 vendas sorteadas entre as opções de categoria.

```python
rng = np.random.default_rng(seed=42) # Usando uma semente fixa para reprodutibilidade
categorias_venda = ['Eletrônicos', 'Roupas', 'Casa']
amostra_vendas = rng.choice(categorias_venda, size=50)
display(amostra_vendas)
```

### Passo 2: Determine o volume de vendas de cada departamento usando `np.unique`.

```python
departamentos, contagens = np.unique(amostra_vendas, return_counts=True)
display(f"Departamentos: {departamentos}")
display(f"Contagens: {contagens}")
```

### Passo 3: Identifique a campanha campeã pelo índice do maior valor.

```python
faturamento_campanhas = np.array([12000, 45000, 23000, 89000, 31000])
indice_campea = np.argmax(faturamento_campanhas)
valor_campea = faturamento_campanhas[indice_campea]

display(f"Faturamento das campanhas: {faturamento_campanhas}")
display(f"Índice da campanha campeã: {indice_campea}")
display(f"Faturamento da campanha campeã: R$ {valor_campea:.2f}")
```
