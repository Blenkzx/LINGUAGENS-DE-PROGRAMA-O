### Nível 1: Básico (URLs, Parâmetros e Cabeçalhos)

Vamos começar com o Exercício 1.1, 1.2, 1.3 e 1.4, que envolvem a criação de uma requisição GET com parâmetros e cabeçalhos, além de verificar o status da resposta.

```python
import requests

# Exercício 1.1: Faça uma requisição GET para a API pública do JSONPlaceholder
url = 'https://jsonplaceholder.typicode.com/posts'

# Exercício 1.2: Utilize o argumento params para filtrar os resultados (por exemplo, buscando apenas o userId igual a 2 ou limitando a quantidade de resultados).
params = {
    'userId': 2, # Filtering by userId
    '_limit': 5  # Limiting the number of results
}

# Exercício 1.3: Na sua requisição, inclua um dicionário de cabeçalhos (headers) passando um User-Agent personalizado com o nome do seu projeto.
headers = {
    'User-Agent': 'MeuProjetoDeRequisicoesPython/1.0' # Custom User-Agent
}

try:
    response = requests.get(url, params=params, headers=headers, timeout=10)

    # Exercício 1.4: Imprima na tela o código de status HTTP e a URL final montada
    print(f"Status Code: {response.status_code}")
    print(f"Final URL: {response.url}")

    # Check if the request was successful (status code 200)
    if response.status_code == 200:
        print("Requisição GET bem-sucedida!")
        # Optional: Print the first part of the JSON response to see the data
        # print("Conteúdo da Resposta (primeiros 500 caracteres):\n", response.text[:500])
    else:
        print(f"Erro na requisição. Status: {response.status_code}")
        response.raise_for_status() # Raises HTTPError for bad responses (4xx or 5xx)

except requests.exceptions.Timeout:
    print("A requisição excedeu o tempo limite.")
except requests.exceptions.RequestException as e:
    print(f"Ocorreu um erro durante a requisição: {e}")
```

### Nível 2: Intermediário (JSON, Erros e Arquivos Binários)

Vamos agora para o **Exercício 2.1**, que consiste em consultar CEPs na API do ViaCEP e armazenar os resultados em um DataFrame do Pandas.

```python
import requests
import pandas as pd

# Exercício 2.1: Crie uma lista com três CEPs diferentes e faça um loop para consultá-los na API do ViaCEP
ceps = ['01001000', '20040030', '70040010'] # Exemplos de CEPs
results = []

print("Consultando CEPs na API do ViaCEP...")
for cep in ceps:
    url = f"https://viacep.com.br/ws/{cep}/json/"
    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status() # Lança um HTTPError para códigos de status de erro (4xx ou 5xx)
        data = response.json() # Utilize o método .json() para converter as respostas
        results.append(data)
        print(f"CEP {cep} consultado com sucesso.")
    except requests.exceptions.Timeout:
        print(f"A requisição para o CEP {cep} excedeu o tempo limite.")
    except requests.exceptions.RequestException as e:
        print(f"Ocorreu um erro ao consultar o CEP {cep}: {e}")

# Armazene os resultados em um DataFrame do Pandas
df_ceps = pd.DataFrame(results)

print("\nDataFrame de CEPs criado:")
print(df_ceps.head())
```

### Nível 2: Intermediário (JSON, Erros e Arquivos Binários) - Continuação

Vamos agora para o **Exercício 2.2** e **2.3**, que abordam o download seguro de arquivos e o salvamento de conteúdo binário.

```python
import requests

# Exercício 2.2: Escreva uma função de download segura utilizando um bloco try/except.
def download_seguro(url, filename, timeout=10):
    try:
        print(f"Tentando baixar {url}...")
        response = requests.get(url, timeout=timeout)
        response.raise_for_status() # Captura erros HTTP (4xx ou 5xx)

        # Exercício 2.3: Salve o conteúdo bruto da resposta em um arquivo local com a extensão .jpg
        with open(filename, 'wb') as f:
            f.write(response.content)
        print(f"Arquivo '{filename}' baixado com sucesso!")
        return True
    except requests.exceptions.HTTPError as e:
        print(f"Erro HTTP ao baixar '{filename}': {e} (Status Code: {response.status_code})")
        return False
    except requests.exceptions.ConnectionError:
        print(f"Erro de conexão ao baixar '{filename}'. Verifique sua internet.")
        return False
    except requests.exceptions.Timeout:
        print(f"A requisição para '{filename}' excedeu o tempo limite.")
        return False
    except requests.exceptions.RequestException as e:
        print(f"Ocorreu um erro inesperado ao baixar '{filename}': {e}")
        return False

# URL para baixar uma imagem aleatória
image_url = 'https://picsum.photos/400/400'
image_filename = 'imagem_aleatoria.jpg'

# Chame a função para baixar a imagem
print("\n--- Executando Exercício 2.2 e 2.3 ---")
download_seguro(image_url, image_filename)

# Exemplo de URL inválida para testar o tratamento de erro (descomente para testar)
# print("\n--- Testando tratamento de erro com URL inválida ---")
# download_seguro('https://picsum.photos/9999/9999', 'imagem_grande_erro.jpg')
```

### Nível 3: Avançado (Webscraping e Ética)

Iniciamos agora os **Exercícios 3.1, 3.2 e 3.3**, focado em ética no web scraping.

```python
import requests

# Exercício 3.1: Antes de fazer a raspagem, faça uma requisição para o arquivo robots.txt do site
# que deseja acessar para verificar as permissões de acesso, demonstrando responsabilidade ética na coleta de dados.

website_url = "https://books.toscrape.com/"
robots_url = website_url + "robots.txt"

try:
    print(f"Verificando robots.txt para: {website_url}")
    response = requests.get(robots_url, timeout=10)
    response.raise_for_status() # Lança um HTTPError para códigos de status de erro

    print("Conteúdo do robots.txt:")
    print(response.text)

    print("\nAnálise básica do robots.txt (verifique as diretivas User-agent e Disallow):")
    if "Disallow" in response.text:
        print("Existem diretivas 'Disallow' neste robots.txt. Consulte-as antes de fazer scraping.")
    else:
        print("Nenhuma diretiva 'Disallow' explícita encontrada, mas ainda é recomendado cautela e respeito aos termos de serviço.")

except requests.exceptions.Timeout:
    print(f"A requisição para {robots_url} excedeu o tempo limite.")
except requests.exceptions.RequestException as e:
    print(f"Ocorreu um erro ao acessar o robots.txt em {robots_url}: {e}")
```

```python
import requests
from bs4 import BeautifulSoup

# Exercício 3.2: Acesse o site voltado para estudos de raspagem https://books.toscrape.com/.
# Utilize a biblioteca BeautifulSoup com o analisador 'html.parser' para localizar os elementos da página.
# Extraia o título e o preço dos 5 primeiros livros utilizando métodos como .find() ou .select().

url_scraping = 'https://books.toscrape.com/'

try:
    print(f"Acessando: {url_scraping} para web scraping...")
    response = requests.get(url_scraping, timeout=10)
    response.raise_for_status() # Lança um HTTPError para códigos de status de erro

    # Cria um objeto BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # Encontra todos os artigos de livro na página
    books = soup.select('article.product_pod')

    print("\nDados dos 5 primeiros livros:\n")
    for i, book in enumerate(books[:5]): # Limita aos 5 primeiros livros
        title = book.h3.a['title']
        price = book.find('p', class_='price_color').text.strip()
        print(f"Livro {i+1}:")
        print(f"  Título: {title}")
        print(f"  Preço: {price}")
        print("\n")

except requests.exceptions.Timeout:
    print(f"A requisição para {url_scraping} excedeu o tempo limite.")
except requests.exceptions.RequestException as e:
    print(f"Ocorreu um erro ao acessar a página para scraping: {e}")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import io

# Re-scraping the data to ensure it's fresh for the DataFrame creation
url_scraping = 'https://books.toscrape.com/'
book_data = []

try:
    print(f"Acessando: {url_scraping} para coletar dados para o CSV...")
    response = requests.get(url_scraping, timeout=10)
    response.raise_for_status() # Lança um HTTPError para códigos de status de erro

    soup = BeautifulSoup(response.text, 'html.parser')
    books = soup.select('article.product_pod')

    for book in books[:5]: # Coleta os 5 primeiros livros
        title = book.h3.a['title']
        price_str = book.find('p', class_='price_color').text.strip() # Ex: 'Â£51.77'
        # Correção: Remova 'Â£' para garantir a conversão correta para float
        price = float(price_str.replace('Â£', ''))
        book_data.append({'Title': title, 'Price': price})

    # Exercício 3.3: Salve os dados extraídos dos livros em um arquivo CSV utilizando o Pandas.
    df_books = pd.DataFrame(book_data)
    csv_filename = 'livros_raspados.csv'
    df_books.to_csv(csv_filename, index=False)

    print(f"\nDados dos livros salvos com sucesso em '{csv_filename}'")
    print("Conteúdo do DataFrame criado:")
    print(df_books.head())

except requests.exceptions.Timeout:
    print(f"A requisição para {url_scraping} excedeu o tempo limite.")
except requests.exceptions.RequestException as e:
    print(f"Ocorreu um erro ao acessar a página para scraping ou salvar o CSV: {e}")
```

```python
import requests
import pandas as pd
import io

# Exercício 3.4: Escolha uma página da Wikipedia que contenha uma tabela de dados.
# Utilize a função pandas.read_html() combinada com io.StringIO() para capturar a tabela da página e transformá-la diretamente em um DataFrame.

# URL da Wikipedia sugerida pelo usuário
wikipedia_url = 'https://en.wikipedia.org/wiki/List_of_countries_by_population'

try:
    print(f"Acessando: {wikipedia_url} para extrair tabelas...")

    # Adicionar um cabeçalho User-Agent para simular um navegador e tentar evitar o erro 403 Forbidden
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    response = requests.get(wikipedia_url, timeout=15, headers=headers) # Aumentar timeout para páginas maiores
    response.raise_for_status() # Lança um HTTPError para códigos de status de erro

    # Usa io.StringIO para tratar o texto HTML como um arquivo para read_html
    # read_html retorna uma lista de DataFrames, um para cada tabela encontrada
    tables = pd.read_html(io.StringIO(response.text))

    print(f"Total de tabelas encontradas na página: {len(tables)}")

    # Geralmente, a tabela principal de dados é a primeira ou uma das primeiras.
    # Inspecione o conteúdo de 'tables' para determinar qual é a tabela correta.
    # Por exemplo, vamos exibir a primeira tabela encontrada.
    if tables:
        df_wikipedia = tables[0] # Pega a primeira tabela
        print("\nPrimeira tabela extraída:")
        display(df_wikipedia.head())

        # Opcional: Salvar a tabela em CSV também
        csv_filename_wiki = 'wikipedia_tabela_populacao.csv'
        df_wikipedia.to_csv(csv_filename_wiki, index=False)
        print(f"\nPrimeira tabela salva em '{csv_filename_wiki}'")
    else:
        print("Nenhuma tabela encontrada na página.")

except requests.exceptions.Timeout:
    print(f"A requisição para {wikipedia_url} excedeu o tempo limite.")
except requests.exceptions.RequestException as e:
    print(f"Ocorreu um erro ao acessar a página ou extrair a tabela: {e}. Embora tenhamos adicionado um User-Agent, a Wikipedia pode ter outras restrições ou a URL pode não conter tabelas no formato esperado.")
except ValueError as e:
    print(f"Erro ao analisar HTML com pandas.read_html(): {e}. Verifique se a página contém tabelas ou a estrutura do HTML. Pode ser que o conteúdo não seja tabular ou esteja em um formato diferente.")
```
