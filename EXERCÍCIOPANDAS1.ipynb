```python
import numpy as np
import pandas as pd

# Base de vendas
dados_vendas = {
    'cliente_id': [101, 102, 103, 101, 104, 102, 105, 103],
    'valor': [3500.75, 189.50, np.nan, 1200.00, 450.00, np.nan, 89.90, 780.50],
    'categoria': [
        'Eletronicos',
        'Livros',
        'Roupas',
        'Eletronicos',
        'Automotivo',
        'Livros',
        'Roupas',
        'Roupas',
    ],
    'data_hora': [
        '2024-01-15 10:23:00',
        '2024-01-18 14:05:00',
        '2024-02-05 09:12:00',
        '2024-02-20 16:40:00',
        '2024-03-02 11:00:00',
        '2024-03-15 18:30:00',
        '2024-04-10 08:20:00',
        '2024-04-22 13:45:00',
    ],
    'status': [
        'Concluído',
        'Concluído',
        'Pendente',
        'Concluído',
        'Cancelado',
        'Concluído',
        'Concluído',
        'Concluído',
    ],
    'email': [
        'maria@gmail.com',
        'joao@outlook.com',
        'ana@yahoo.com',
        'maria@gmail.com',
        'carlos@gmail.com',
        'joao@outlook.com',
        'lucas@empresa.com.br',
        'ana@yahoo.com',
    ],
}

# Base de clientes
dados_clientes = {
    'cliente_id': [101, 102, 103, 104, 105],
    'nome': [
        'Maria Silva',
        'Joao Souza',
        'Ana Oliveira',
        'Carlos Lima',
        'Lucas Mendes',
    ],
    'cidade': [
        'Sao Paulo',
        'Rio de Janeiro',
        'Belo Horizonte',
        'Curitiba',
        'Salvador',
    ],
}

df_vendas = pd.DataFrame(dados_vendas)
df_clientes = pd.DataFrame(dados_clientes)
```

## Parte 1: Diagnóstico e Limpeza (Básico)

```python

print("Dimensões de df_vendas:", df_vendas.shape)
print("\nInformações de df_vendas:")
df_vendas.info()
```

```python


df_vendas['valor'] = df_vendas.groupby('categoria')['valor'].transform(
    lambda x: x.fillna(x.median() if not x.isnull().all() else df_vendas['valor'].median())
)

print("\nValores nulos após o preenchimento:")
display(df_vendas[df_vendas['valor'].isnull()])
display(df_vendas.head())
```

```python

transacoes_filtradas = df_vendas.query("status == 'Concluído' and valor > 500.00")

print("\nTransações com status 'Concluído' e valor superior a R$ 500,00:")
display(transacoes_filtradas)
```

## Parte 2: Cruzamento e Transformação (Intermediário)

```python

df_merged = pd.merge(df_vendas, df_clientes, on='cliente_id', how='left')

print("\nDataFrame após a junção:")
display(df_merged.head())
```

```python

df_merged['media_categoria'] = df_merged.groupby('categoria')['valor'].transform('mean')

print("\nDataFrame com a coluna 'media_categoria':")
display(df_merged.head())
```

```python

clientes_gmail_count = df_merged['email'].str.contains('@gmail.com', na=False).sum()

print(f"\nNúmero de clientes que usam @gmail.com: {clientes_gmail_count}")
```

## Parte 3: Análise Temporal e Agregação (Intermediário)

```python

df_merged['data_hora'] = pd.to_datetime(df_merged['data_hora'])
df_merged['mes_extenso'] = df_merged['data_hora'].dt.month_name()
df_merged['dia_da_semana'] = df_merged['data_hora'].dt.day_name()

print("\nDataFrame com colunas de tempo extraídas:")
display(df_merged[['data_hora', 'mes_extenso', 'dia_da_semana']].head())
```

```python

pivot_table_vendas = pd.pivot_table(
    df_merged,
    values='valor',
    index='categoria',
    columns='cidade',
    aggfunc='sum',
    fill_value=0
)

print("\nTabela Dinâmica de Vendas por Categoria e Cidade:")
display(pivot_table_vendas)
```
